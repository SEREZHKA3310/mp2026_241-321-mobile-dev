```
// плохие именаа
int w = 20;
int h = 5;
char s = '*';
bool f = true;

int width = 20;
int height = 5;
char symbol = '*';
bool isFilled = true; // Python: is_filled

// (1)
for (int r = 0; r < height; r++)
{
  for (int c = 0; c < width; c++)
  {
    if (isFilled)
    {
      Console.Write(symbol);
    }
    else
    {  
      if (r == 0 || r == height - 1 || c == 0 || c == width - 1)
      {
        Console.Write(symbol);
      }
      else
      {
        Console.Write(' ');
      }
    }
  }
  Console.WriteLine();
}
```

```
int width = 20;
int height = 5;
char symbol = '*';
bool isFilled = !true;

// (2)
for (int r = 0; r < height; r++)
{
  for (int c = 0; c < width; c++)
  {
    bool isSymb = isFilled || r == 0 || r == height - 1 || c == 0 || c == width -1;
    Console.Write(isSymb ? symbol : ' ');
  }
  Console.WriteLine();
}
```

```
int width = 20;
int height = 5;
char symbol = '*';
bool isFilled = !true;
char symbolClear = ' ';

// (3)
for (int r = 0; r < height; r++)
{
  if (isFilled || r == 0 || r == height - 1)
    Console.WriteLine(new string(symbol, width));
  else
    Console.WriteLine(symbol + new string(symbolClear, width - 2) + symbol);
}
```

```
// (4)
/// <summary>
/// Нарисовать прямоугольник в консоль
/// </summary>
/// <param name="width">Ширина фигуры</param>
/// <param name="height">Высота фигуры</param>
/// <param name="symbol">Символ рисования</param>
/// <param name="isFilled">Признак залития фигуры</param>
/// <param name="symbolClear">Символ очистки</param>
void PrintRectangle
(
  int width,
  int height,
  char symbol = '*',
  bool isFilled = true,
  char symbolClear = ' '
)
{
  for (int r = 0; r < height; r++)
  {
    if (isFilled || r == 0 || r == height - 1)
      Console.WriteLine(new string(symbol, width));
    else
      Console.WriteLine(symbol + new string(symbolClear, width - 2) + symbol);
  }
}

PrintRectangle(25, 4);
PrintRectangle(25, 4, symbol:'@', isFilled: false, symbolClear: '-');
```

```
// (5)
/// <summary>
/// Нарисовать прямоугольник в консоль
/// </summary>
/// <param name="width">Ширина фигуры</param>
/// <param name="height">Высота фигуры</param>
/// <param name="symbol">Символ рисования</param>
/// <param name="isFilled">Признак залития фигуры</param>
/// <param name="symbolClear">Символ очистки</param>
string[] GetRectangle
(
  int width,
  int height,
  char symbol = '*',
  bool isFilled = true,
  char symbolClear = ' '
)
{
  List<string> result = new();
  for (int r = 0; r < height; r++)
  {
    if (isFilled || r == 0 || r == height - 1)
      Console.WriteLine(new string(symbol, width));
    else
      Console.WriteLine(symbol + new string(symbolClear, width - 2) + symbol);
  }
  return result.ToArray();
}

var shape = GetRectangle(25, 4);
Console.WriteLine(string.Join(Environment.NewLine, shape));
```
