```
/// <summary>
/// Нарисовать равнобедренный треугольник в консоль
/// </summary>
/// <param name="width">Ширина основания фигуры</param>
/// <param name="height">Высота фигуры</param>
/// <param name="symbol">Символ рисования</param>
/// <param name="isFilled">Признак залития фигуры</param>
/// <param name="symbolClear">Символ очистки</param>
string[] GetTriangle
(
  int width,
  int height,
  char symbol = '*',
  bool isFilled = true,
  char symbolClear = ' '
)
{
  List<string> result = new();

  for (int r = 0; r < height; r++)
  {
    // Вычисляем ширину текущего слоя (от 1 на вершине до width у основания)
    int lineWidth = height > 1 
      ? (int)Math.Round(1.0 + (double)r * (width - 1) / (height - 1))
      : width;

    // Центрируем строку с помощью отступов слева и справа
    int padding = (width - lineWidth) / 2;
    string padStr = new string(symbolClear, padding);

    string inner;
    if (isFilled || r == 0 || r == height - 1)
    {
      inner = new string(symbol, lineWidth);
    }
    else if (lineWidth <= 2)
    {
      inner = new string(symbol, lineWidth);
    }
    else
    {
      inner = symbol + new string(symbolClear, lineWidth - 2) + symbol;
    }

    result.Add(padStr + inner + padStr);
  }

  return result.ToArray();
}

var shape = GetTriangle(25, 9, '*', false);
Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
/// <summary>
/// Нарисовать зебру
/// </summary>
/// <param name="width">Ширина фигуры</param>
/// <param name="height">Высота фигуры</param>
/// <param name="stripeHeight">Толщина полосы</param>
/// <param name="symbol">Символ рисования</param>
/// <param name="symbolClear">Символ очистки</param>
string[] GetZebra
(
    int width,
    int height,
    int stripeHeight = 1,
    char symbol = '*',
    char symbolClear = ' '
)
{
    List<string> result = new();

    for (int r = 0; r < height; r++)
    {
        if ((r / stripeHeight) % 2 == 0)
            result.Add(new string(symbol, width));
        else
            result.Add(new string(symbolClear, width));
    }

    return result.ToArray();
}

var shape = GetZebra(20, 8, 2);

Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
/// <summary>
/// Нарисовать крест
/// </summary>
/// <param name="size">Размер фигуры</param>
/// <param name="symbol">Символ рисования</param>
string[] GetPlus
(
    int size,
    char symbol = '*'
)
{
    List<string> result = new();

    int center = size / 2;

    for (int r = 0; r < size; r++)
    {
        if (r == center)
            result.Add(new string(symbol, size));
        else
            result.Add(
                new string(' ', center) +
                symbol +
                new string(' ', size - center - 1)
            );
    }

    return result.ToArray();
}

var shape = GetPlus(9);
Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
/// <summary>
/// Нарисовать пирамиду
/// </summary>
/// <param name="height">Высота фигуры</param>
/// <param name="symbol">Символ рисования</param>
/// <param name="isFilled">Признак залития фигуры</param>
string[] GetPyramid
(
    int height,
    char symbol = '*',
    bool isFilled = false
)
{
    List<string> result = new();

    for (int r = 0; r < height; r++)
    {
        int spaces = height - r - 1;
        int width = r * 2 + 1;

        string line = new string(' ', spaces);

        if (isFilled)
        {
            line += new string(symbol, width);
        }
        else
        {
            if (r == 0)
                line += symbol;
            else if (r == height - 1)
                line += new string(symbol, width);
            else
                line += symbol +
                        new string(' ', width - 2) +
                        symbol;
        }

        result.Add(line);
    }

    return result.ToArray();
}

var shape = GetPyramid(6);

Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
/// <summary>
/// Нарисовать песочные часы
/// </summary>
/// <param name="height">Высота фигуры</param>
/// <param name="symbol">Символ рисования</param>
/// <param name="isFilled">Признак залития фигуры</param>
string[] GetHourglass
(
    int height,
    char symbol = '*',
    bool isFilled = true
)
{
    List<string> result = new();

    int width = height * 2 - 1;

    for (int r = 0; r < height; r++)
    {
        int spaces = Math.Min(r, height - r - 1);
        int currentWidth = width - spaces * 2;

        string line = new string(' ', spaces);

        if (isFilled)
        {
            line += new string(symbol, currentWidth);
        }
        else
        {
            if (currentWidth == 1)
                line += symbol;
            else
                line += symbol +
                        new string(' ', currentWidth - 2) +
                        symbol;
        }

        result.Add(line);
    }

    return result.ToArray();
}

var shape = GetHourglass(7);

Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
/// <summary>
/// Нарисовать стрелку
/// </summary>
/// <param name="height">Высота фигуры</param>
/// <param name="symbol">Символ рисования</param>
/// <param name="pointUp">Направление стрелки вверх</param>
string[] GetArrow
(
    int height,
    char symbol = '*',
    bool pointUp = true
)
{
    List<string> result = new();

    int headHeight = height / 2;
    int width = headHeight * 2 + 1;

    for (int r = 0; r < height; r++)
    {
        string line = "";

        int currentRow = pointUp
            ? r
            : height - r - 1;

        if (currentRow < headHeight)
        {
            int spaces = headHeight - currentRow;
            int symbols = currentRow * 2 + 1;

            line += new string(' ', spaces);
            line += new string(symbol, symbols);
        }
        else
        {
            int spaces = headHeight - 1;

            line += new string(' ', spaces);
            line += new string(symbol, 3);
        }

        result.Add(line);
    }

    return result.ToArray();
}

var shape = GetArrow(9);

Console.WriteLine(string.Join(Environment.NewLine, shape));
```
