```
byte x = 3;
byte y = 5;
Console.WriteLine($"{x:B4} = x");
Console.WriteLine($"{y:B4} = y");
Console.WriteLine();
Console.WriteLine($"{x & y:B4} - AND");
Console.WriteLine($"{x | y:B4} - OR");
Console.WriteLine($"{x ^ y:B4} - XOR");
Console.WriteLine($"{x << 1:B4} = x << 1");
Console.WriteLine($"{x >> 1:B4} = x >> 1");
```

```
var z = 3;
Console.WriteLine($"{z:B32}");
Console.WriteLine($"{~z:B} - NOT");
Console.WriteLine($"{(byte)~z:B} - NOT");
```

```
for (int i = 0; i < 8; i++)
{
  var r = 1 << i;
  Console.WriteLine($"1 << {i} = {r, 3} | {r:88}");
}
```

```
// Q: Как проверить на бит по маске?
byte n = 0b_0001_0111;
byte checkBit = 0b_0001_1000;
Console.WriteLine($"{n & checkBit:B8}");
Console.WriteLine((n & checkBit) == checkBit);
```

```
// Q: Как проверить бит по номеру?
//          7654 3210
byte n = 0b_0001_0111;
byte numBit = 3;
Console.WriteLine($"{1 << numBit:B8}");
Console.WriteLine((n & 1 << numBit) != 0);
```
