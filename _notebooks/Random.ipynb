<h1>Random</h1>

```
Random rnd = new();
```

```
Console.WriteLine(rnd.Next(10));
Console.WriteLine(rnd.Next(-10, -5));
Console.WriteLine(rnd.Next());
Console.WriteLine(rnd.NextInt64());
Console.WriteLine(rnd.NextDouble()); // [0.0; 1.0]
```

```
var numbers = new int[10];
for (int i = 0; i < 10; i++)
  numbers[i] = rnd.Next(100);
numbers
```

```
Random rndFix = new(100);

var numbers = new int[10];
for (int i = 0; i < 10; i++)
  numbers[i] = rndFix.Next(100);
numbers
```
