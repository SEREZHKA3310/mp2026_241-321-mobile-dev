<h1>Функции (Методы)</h1>

```
void PrintString()
{
  Console.WriteLine("Строка");
}
PrintString();
```

```
int Sum1(int a, int b)
{
  return a + b;
}
int Sum2(int a, int b) => a + b;

// Позиционные аргументы
Console.WriteLine(Sum1(1, 2));
Console.WriteLine(Sum1(10, 20));

// Именованные аргументы
Console.WriteLine(Sum1(1, b:2));
Console.WriteLine(Sum1(b:1, a:2));
```

```
int Sum3(int a, int b = 10, int c = 100) => a + b + c;
Console.WriteLine(Sum3(1));
Console.WriteLine(Sum3(1, 2));
Console.WriteLine(Sum3(1, 2, 3));
```

```
void Printparams1(int[] values)
{
  Console.WriteLine(string.Join(" | ", values));
}
Printparams1(new int[] {1, 2, 3, 4});
Printparams1([1, 2, 3, 4]);
```

```
// Методы расширения
static void Print(this string source)
{
  Console.WriteLine(source);
}
static void Print(this string source, string border)
{
  Console.WriteLine($"{border}{source}{border}");
}
static void PrintUpper(this string source)
{
  Console.WriteLine(source.ToUpper());
}

string st = "Строка";
st.Print();
st.Print("*");
st.Print("##");
st.PrintUpper();
```
