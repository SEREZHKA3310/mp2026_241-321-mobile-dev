```
enum Keys : byte
{
    None = 0, // 0b_0000_0000
    Red = 1, // 0b_0000_0001 | 1 << 0
    Green = 2, // 0b_0000_0010 | 1 << 1
    Blue = 4, // 0b_0000_0100 | 1 << 2
    Black = 1 << 3, // 0b_0000_1000 | 1 << 3
    RedGreen = Red | Green
}

Keys keys;
Console.WriteLine(keys);

keys |= Keys.Green; // добавить
Console.WriteLine(keys);

keys |= Keys.Red; // добавить
Console.WriteLine(keys);

keys ^= Keys.Red; // удалить
Console.WriteLine(keys);
```
